## Ejercicio 16
Crea un array cuadrado de 2D, con las dimensiones que quieras, pero que todo el borde esté compuesto por 1s, y el interior por 0s

````python
import numpy as np

x = np.ones((5,5))

x[1:-1, 1:-1] = 0

print(x)
````

````python
array_16 = np.ones(25).reshape(5,5)

array_16[1:4, 1:4] = 0

print(array_16)
````

````python
array_16 = np.zeros(25).reshape(5,5)

array_16[[0, -1], :] = 1
array_16[:, [0, -1]] = 1

print(array_16)
````

````python
array_16 = np.zeros((5,5))

array_16[0:10:9] = 1
array_16.T[0:10:9] = 1

print(array_16)
````

````python
array_16 = np.full(25,0).reshape(5,5)

array_16[0:5,0] = 1
array_16[0] = 1
array_16[0:5,4] = 1
array_16[4] = 1

print(array_16)
````
