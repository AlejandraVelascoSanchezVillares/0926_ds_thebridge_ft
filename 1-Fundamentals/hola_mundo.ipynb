```python
def presentar(nombre, edad, ciudad='Madrid'):
    return f'{nombre}, {edad} años, de {ciudad}'

datos = {'nombre': 'Lucía', 'edad': 30, 'ciudad': 'Sevilla'}

presentar(**datos)
# 'Lucía, 30 años, de Sevilla'
```
