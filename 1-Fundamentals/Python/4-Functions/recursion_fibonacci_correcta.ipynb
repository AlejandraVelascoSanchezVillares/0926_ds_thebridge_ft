## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def convertidor(entrada):
    lista = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
    if entrada in range(7):
        return dict(enumerate(lista))[entrada-1]
    else:
        return "El input debe ser un número del 1 al 7"

convertidor(4)
````

````python
def convertidor(entrada):
    diccionario = {1:"Lunes", 2:"Martes", 3:"Miércoles", 4:"Jueves", 5:"Viernes", 6:"Sábado", 7:"Domingo"}
    if entrada in range(7):
        return diccionario[entrada]
    else:
        return "El input debe ser un número del 1 al 7"

convertidor(4)
````

````python
def convertidor(entrada):
    if entrada == 1:
        salida = "Lunes"
    elif entrada == 2:
        salida = "Martes"
    elif entrada == 3:
        salida = "Miércoles"
    elif entrada == 4:
        salida = "Jueves"
    elif entrada == 5:
        salida = "Viernes"
    elif entrada == 6:
        salida = "Sábado"
    elif entrada == 7:
        salida = "Domingo"
    else:
        salida = "El input debe ser un número del 1 al 7"
    return salida

convertidor(6)
````

## Ejercicio 2
En el ejercicio 8 de bucles, creábamos una pirámide invertida, cuyo número de pisos venía determinado por un input del usuario. Crear una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide, es decir, elimina la sentencia input.

````python
def inv_piramide(pisos):
    for j in range(pisos):
        print(*list(i+1 for i in range(pisos-j)))

inv_piramide(9)
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea la letra a contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayuscula, como minúscula

````python

def contador(texto:str, letra:str):
    if type(texto) != str and (type(letra) != str):
        return("El input debe ser 2 strings")
    contador = texto.lower().count(letra.lower())
    return contador

contador("VeamOs aHORa","o")
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
def diccionario(texto):
    diccionario = {}
    if type(texto) != str:
        return("El input debe ser un string")
    else:
        for letra in texto:
            diccionario[letra] = texto.count(letra)
    return diccionario

print(diccionario("qwerttyuiiuyttyuiiuiuyuyu"))
````

````python
def diccionario(texto):
    allet = {}

    for i in texto.lower():
        if i in allet:
            allet[i] = allet[i] + 1

        else:
            allet[i] = 1
    return allet

diccionario("en esta clase aprendemos mucho Python")
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
lista1 = ["hola", 9, 219, "adios"]
comando1 = "add"
elemento1 = "hola"

def aplicar(lista, comando, elemento):
    if elemento in lista:
        return getattr(lista, comando)(elemento)
    else:
        return lista

aplicar(lista1, "remove", 4)
print(lista1)
````

````python
lista1 = ["hola", 9, 219, "adios"]
comando1 = "add"
elemento1 = "hola"

def modificar(lista, comando, elemento = None):
    if comando == "add":
        lista.append(elemento)
    elif comando == "remove":
        lista.remove(elemento)
    else:
        print("El comando es un string")
    return lista

print(modificar(lista1, comando1, elemento1))
````

## Ejercicio 8

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(cota, contador = 1, a = 0, b = 1):
    c = a + b
    print(c)
    if contador >= cota:
        return c
    contador += 1
    fibonacci(cota, contador, b, c)

fibonacci(9)



# 0 + 0 = 0 ---> sum + ant = 0
# 0 + 1 = 1 ---> sum + ant = 1
# 1 + 1 = 2 ---> sum + ant = 2
# 1 + 2 = 3 ---> sum + ant = a
# 2 + 3 = 5 ---> ant +   a = b
# 3 + 5 = 8 --->   a +   b = c
# 5 + 8 = 13 -->   b +   c = d
# 8 + 13 = 21 ->   c +   d = e
````

Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el area de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
from math import pi

def cuadrado(lado):
    return lado ** 2

def triangulo(base, altura):
    return base * altura / 2

def circulo(r):
    return pi * (r ** 2)
````

````python
print(2 * circulo(10) + triangulo(3,7))
print(cuadrado(10) + circulo(4) + 2 * circulo(6) + 5 * triangulo(2,4))
````
