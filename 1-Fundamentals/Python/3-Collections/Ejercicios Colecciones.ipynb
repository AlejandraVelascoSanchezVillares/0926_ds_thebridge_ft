## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']

````python
ejer_3_1 = ["Hola", "amigo"]

ejer_3_2 = ["Que", "tal"]

ejer_3 = []

for i in ejer_3_1:
    for j in ejer_3_2:
        ejer_3.append(i + " " + j)

print(ejer_3)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

Consejo: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

longitud = len(ejer_5)

while longitud > 0:
    for i in ejer_5:
        if i != 3:
            continue
        else:
            ejer_5.remove(i)
    longitud = longitud - 1
    
print(ejer_5)
````

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

for i in ejer_5[:]:
    if i == 3:
        ejer_5.remove(i)
    
print(ejer_5)
````

## Ejercicio 16
Multiplica todos los valores del diccionario
> ```ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

producto = 1

for i in ejer_16:
    producto = producto * ejer_16[i]

print(producto)
````
